```python
import pandas as pd

pasta = "/content/drive/MyDrive/DataLake/processados/clima"
arquivo = f"{pasta}/rn_2007_2025_completo.csv"

df_dados = pd.read_csv(arquivo)
```

```python
df_dados["time"] = pd.to_datetime(df_dados["time"])

df_dados = df_dados[
    (df_dados["time"].dt.year >= 2015) &
    (df_dados["time"].dt.year <= 2025)
].copy()

df_dados["mes"] = df_dados["time"].dt.month
df_dados["ano"] = df_dados["time"].dt.year

grupos = df_dados.groupby(
    ["municipio", "codigo_ibge", "ano", "mes"]
)

df_dados_mes_municipio = grupos["temperature_2m_mean"].mean()
df_dados_mes_municipio = df_dados_mes_municipio.reset_index()

df_dados_mes_municipio["temperature_2m_mean"] = (df_dados_mes_municipio["temperature_2m_mean"].round(2))

df_dados_mes_municipio = df_dados_mes_municipio.rename(columns={"temperature_2m_mean": "temperatura_media"})

df_dados_mes_municipio.to_csv(f"{pasta}/temperatura_mensal_municipio_rn_2015_2025.csv", index=False)
```
