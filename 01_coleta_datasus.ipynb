#Coleta dos dados do DATASUS
Os dados coletados são secundários baixados automaticamente por FTP, disponibilizados pelo Quantílica - [Acidente por Animais Peçonhentos](https://quantilica.com/alpaca/dados/sinan-anim#arquivos)

A base reúne notificações de acidentes com animais peçonhentos registradas no SINAN, disponibilizadas pelo DATASUS. O Quantilica organiza os arquivos de 2007 a 2026, com informações sobre data, município e tipo de animal. O Quantílica reúne os arquivos em formato DBC e disponibiliza um script para baixá-los automaticamente. Após a coleta, transferimos os arquivos para um DataLake no Google Drive.
